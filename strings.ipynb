* s [-n]  Muestra el digito que corresponde a la posicion indicada, comenzando a leer de derecha a izquierda.

```python
s = 'Spam'
s[len(s)-1]
```

* s[len(s)-1]  Tambien muestra el ultimo digito que corresponde a la posicion indicada. Primero indica la longitud de la cadena con len (s) y con -1 indica la posicion. len(s)= 4 -1= posicion 3

```python
s = "Spam"

s [1:3]
```

* El **slicing** sirve para coger una parte de una cadena. Al coger ***s [1:3]*** devuelve lo que hay entre la posicion 1 y 3 pero no incluye la posicion 3.

```python
s = "Spam"

s [1::]
```

* ***s [1::]*** Devuelve desde la posicion 1 (incluida) hasta el final.

```python
s = "Spam"
s[:-1]
```

* ***s[:-1]*** Devuelve todo menos la ultima letra.

```python
s = "Spam"
s[:]
```

* ***s[:]*** Devuelve toda la cadena

```python
s = "Spam"
s[::-1]
```

* ***s[::-1]*** Devuelve la cadena invertida. **s[inicio:final:paso]**

```python
s = "Spam"
s + "eggs"
```
